# Getting Started with Market Risk VaR System

This notebook demonstrates the basic usage of the VaR calculation system.

```python
import sys
sys.path.append('..')

from src.data.data_collector import DataCollector
from src.data.preprocessor import DataPreprocessor
from src.models.var_calculator import VaRCalculator
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
```

## 1. Fetch Market Data

```python
# Initialize data collector
collector = DataCollector()

# Fetch Apple stock data
ticker = 'AAPL'
data = collector.fetch_stock_data(ticker, period='2y')
print(f"Downloaded {len(data)} days of data for {ticker}")
data.tail()
```

## 2. Calculate Returns

```python
# Prepare returns data
preprocessor = DataPreprocessor()
returns_data = preprocessor.prepare_returns_data(data)
returns = returns_data['Returns'].dropna()

# Plot returns
plt.figure(figsize=(12, 6))
returns.plot()
plt.title(f'{ticker} Daily Returns')
plt.ylabel('Returns')
plt.grid(True)
plt.show()
```

## 3. Calculate VaR

```python
# Initialize VaR calculator
var_calc = VaRCalculator(confidence_level=0.95)
position_value = 1_000_000

# Calculate VaR using all methods
results = var_calc.calculate_all_methods(returns, position_value)
print("\nVaR Calculations:")
print(results)
```

## 4. Visualize Results

```python
# Plot VaR comparison
plt.figure(figsize=(10, 6))
plt.bar(results['Method'], results['VaR'])
plt.title('VaR Comparison Across Methods')
plt.ylabel('VaR (USD)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
